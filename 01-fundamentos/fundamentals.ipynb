## Tipos de Variáveis em Python

Uma variável é um objeto, afinal tudo é um objeto em Python

int. --> Inteiro

string --> texto

float --> números com casas decimais (ponto flutuante)

bool ou boolean --> True ou False

```python
faturamento = 1000
type(faturamento)
```

```python
faturamento = 1000.00
type(faturamento)
```

```python
faturamento = '1.000'
type(faturamento)
```

```python
ganha_bonus = True
type(ganha_bonus)
```

```python
faturamento = 2000
custo = 500
lucro = faturamento - custo

print('O faturamento da loja foi ' + str(faturamento))
print('O custo da loja foi ' + str(custo))
print('O lucro da loja foi ' + str(lucro) + ' reais')
```

```python
# Método Format

faturamento = 2000
custo = 500
lucro = faturamento - custo

print('O faturamento da loja foi {}. O Custo da loja foi {}. Assim, o lucro da loja foi {}'.format(faturamento, custo, lucro))


```
