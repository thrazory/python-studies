# Comparadores e Operadores Lógicos em Python

Os comparadores são utilizados para estabelecer relações entre valores, resultando sempre em verdadeiro (`True`) ou falso (`False`).

## Tabela de Comparadores

- `==`: Igual a
- `!=`: Diferente de
- `>`: Maior que
- `>=`: Maior ou igual a
- `<`: Menor que
- `<=`: Menor ou igual a
- `in`: Verifica se um texto existe dentro de outro
- `not`: Inverte o resultado da comparação (negação)

> **Dica:** Quando precisar criar uma estrutura condicional onde um dos caminhos não deve executar nenhuma ação, utilize a palavra reservada `pass`.

## Exemplos Práticos

### Exemplo 1: Igualdade entre valores

```python
faturamento_loja_1 = 2500
faturamento_loja_2 = 2200

if faturamento_loja_1 == faturamento_loja_2:
    resultado = 'Os faturamentos são iguais'
else:
    resultado = 'Os faturamentos são diferentes'

print(f'{resultado}')
```

### Exemplo 2: Diferença

```python
email = 'lira@gmail.com'

if email != 'lira@gmail.com':
    mensagem = 'Esse não é o e-mail do Lira'
else:
    mensagem = 'E-mail correto'

print(f'{mensagem}')
```

### Exemplo 3: Uso de `not` e `pass`

```python


email_usuario = 'usuario@gmail.com'

if not '@' in email_usuario:
    print('E-mail inválido')
else:
    pass
```
