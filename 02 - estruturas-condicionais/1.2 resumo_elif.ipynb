# Estrutura Condicional: elif

Quando precisamos testar mais de duas possibilidades (além do clássico sim/não do `if` e `else`), utilizamos o `elif` (abreviação de *else if*).

## Estrutura Básica

```python
if condicao_1:
    # Executado se a condicao_1 for verdadeira
elif condicao_2:
    # Executado se a condicao_1 for falsa e a condicao_2 for verdadeira
else:
    # Executado se todas as condições anteriores forem falsas
```

## Exemplo Pratico: Calculo de Bonus de Vendas

Vamos avaliar o bônus de um funcionário com base na sua meta de vendas (meta de R$ 20.000):

- Abaixo da meta: Sem bônus (0%)
- Acima da meta: Bônus de 3%
- Mais do que o dobro da meta (acima de R$ 40.000): Bônus de 7%

````python
meta = 20000
vendas = 15000

if vendas > (2 * meta):
    bonus = vendas * 0.07
elif vendas > meta:
    bonus = vendas * 0.03
else:
    bonus = 0

print(f'O bônus foi de R$ {bonus:.2f}')
````
