#  Blocos e Indentação em Python

Em Python, a indentação não serve apenas para deixar o código bonito: **ela define onde os blocos de código começam e terminam** (substituindo chaves `{}` de outras linguagens).

--- 
### 1. Várias ações dentro de um bloco
Quando uma condição é verdadeira, podemos executar múltiplas linhas de código, desde que todas estejam corretamente indentadas:

```python
if condicao:
    # Ação 1
    # Ação 2
else:
    # Ação alternativa
```

--- 
### 2. Exemplo Prático: Fundo de Investimentos
Vamos avaliar taxas de performance com base no rendimento da carteira:
- Menor ou igual a 5%: Sem taxa ($0$)
- Entre 5% e 20%: Taxa de 2% ($0.02$)
- Acima de 20%: Taxa de 4% ($0.04$)

````python
meta = 0.05
rendimento = 0.02

if rendimento > meta:
    if rendimento > 0.20:
        taxa = 0.04
    else:
        taxa = 0.02
else:
    taxa = 0.0

print(f'A taxa foi de {taxa}')
````

--- 
###  Dica de Ouro: Evite Repetição de Código
Note que no código original o comando `print()` era repetido dentro de cada `if/else`. Usar uma variável auxiliar (como `taxa`) e imprimir o resultado **apenas uma vez no final** deixa o código muito mais limpo e fácil de manter.
